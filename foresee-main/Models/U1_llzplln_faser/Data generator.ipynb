```python
xHs = [-2, -1.2, 0, 2]
modes_decay = ["Ne_Ne","e_e", "mu_mu", "tau_tau", "nu_nu", "hadrons"]

mN1, mN2, mN3 = 1/3, 1/3, 1000
    # coupling = np.linspace(1e-11, 1e-5, 1000)
name = "U1_X_phi4"
modelD = darkcast.Model(name, m1 = mN1, m2 = mN2, m3 = mN3)

mass = np.logspace(-3, 1, 10000)
for mode in modes_decay:
    for xH in xHs:
        if mode == "Ne_Ne":
            f = open("model/br_RHN3_phi4/"+str(xH)+"/"+"rhn"+".txt", "w")
        else:
            f = open("model/br_RHN3_phi4/"+str(xH)+"/"+mode+".txt", "w")
        bfracs = []
        for m in mass:
            if mode == "nu_nu":
                br = modelD.bfrac("neutrinos", m, xH)
            elif mode == "hadrons": 
                br = modelD.bfrac("hadrons", m, xH)
            else:
                br = modelD.bfrac(mode, m, xH)
            f.write(str(m)+"   "+str(br)+"\n")
        f.close()
```
