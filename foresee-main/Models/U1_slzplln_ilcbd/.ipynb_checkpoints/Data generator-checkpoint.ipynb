```python
xHs = [-2, -1, -0.5, 0, 0.5, 1, 2]

mN1, mN2, mN3 = 1/3, 1/3, 1000
    # coupling = np.linspace(1e-11, 1e-5, 1000)
name = "U1_X_phi4"
modelD = darkcast.Model(name, m1 = mN1, m2 = mN2, m3 = mN3)

mass = np.logspace(-3, 1, 10000)
for xH in xHs:
    f = open("model/br_RHN3_phi4/"+str(xH)+"/"+"rhn"+".txt", "w")
    bfracs = []
    for m in mass:
        br = modelD.bfrac("Ne_Ne", m, xH)
        f.write(str(m)+"   "+str(br)+"\n")
    f.close()
```

```python
xHs = [-2,-1, -0.5, 0, 0.5, 1, 2]
mN = 1/3
mass = np.logspace(-3, 1, 10000)
name = "U1_X_phi4"
states = ndec.states_all()
for xH in xHs:
    f = open("model/"+str(xH)+"ctau_RHN3_phi4.txt", "w")
    ctaus = []
    for m in mass:
    # coupling = np.linspace(1e-11, 1e-5, 1000)
        tau = 0
        for state in states:
            tau += ndec.Nwidth(m*mN, xH, state, 1, 1)*ndec.factor(state)
        tau = 1/tau
        ctau = 3e8 * tau
        f.write(str(m)+"   "+str(ctau)+"\n")
    f.close()
```

```python
xHs = [-2, -1, -0.5, 0, 0.5, 1, 2]
mN = 1/3
statesN = ndec.states_decay()
mass = np.logspace(-3, 1, 10000)
for xH in xHs:
    f = open("model/br_RHN3_phi4/"+str(xH)+"/"+"visible"+".txt", "w")
    bfracs = []
    for m in mass:
        br = 0
        for state in statesN.keys(): 
            br+= ndec.NBR(m*mN, xH, statesN[state], 1, 1)
        f.write(str(m)+"   "+str(br)+"\n")
    f.close()
```
